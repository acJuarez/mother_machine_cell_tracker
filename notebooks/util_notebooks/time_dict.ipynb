### from image_segmentation_kymographs google sheets will generate time_dict for everything provided

```python
path ="/Users/adrianjuarez/Downloads/image_segmentation_kymographs.csv"
meta_data = pd.read_csv(path)
```

```python
meta_data_filtered = meta_data.dropna(subset='Start_frame')
```

```python
import json
import pandas as pd

def create_time_dict(df):
    time_dict = {}
    
    for _, row in df.iterrows():
        experiment = row['experiment_name']
        fov = str(row['FOV']).zfill(3)  # Convert to 3-digit string with leading zeros
        trench = str(row['trench_id'])
        start = int(row['Start_frame'])  # Convert to int
        end = int(row['End_frame']) if pd.notna(row['End_frame']) else None  # Convert to int if not null
        
        # Create nested structure
        if experiment not in time_dict:
            time_dict[experiment] = {}
        if fov not in time_dict[experiment]:
            time_dict[experiment][fov] = {}
        
        time_dict[experiment][fov][trench] = {
            "start": start,
            "end": end
        }
    
    return time_dict

# Use the function
time_dict = create_time_dict(meta_data_filtered)

# Convert to JSON string without spaces
time_dict_string = json.dumps(time_dict, separators=(',', ':'))
print(time_dict_string)
```
